# Atributos GIS
#### Objetivos:
- Recorrer carpetas y encontrar todos los archivos .shp
- Abrirlos y escribir los atributos
- Guardar en una variable el nombre
- Guardar en una variable el tipo de dato que es (line, poligono,etc)

```python
# Librerias
from pathlib import Path
import geopandas as gpd
import pandas as pd
# import fiona
import pyogrio

print("librerias ok")
```

```python

#carpeta = Path(r"E:/HYTSA/15 - HYTSA - C594 EL SALV DISEÑOS MAJAGÛE/00_Capas")
# carpeta = Path(r"E:/HYTSA/12 - HYTSA - C565 SAL  DISEÑO DE SANEAMIENTO Y AGUA POTABLE DESTINOS TURÍSTICOS/03_Datos")
carpeta = Path(r"E:/Luciano/00-Trabajos/00 - HYTSA/04 - HYTSA - El Salvador y Honduras")
df_rbind = []

for file in carpeta.rglob("*.shp"):

    print(file)

    try:
        # Leer shapefile
        gdf = gpd.read_file(file)

        # Atributos
        atributos = gdf.columns.tolist()

        # Tipo de geometría
        tipo = gdf.geometry.geom_type.unique().tolist()

        df = pd.DataFrame([{
            "archivo": file.name,
            "path": str(file),
            "tipo_geometria": tipo,
            "atributos": atributos,
            "estado": "OK",
            "error": None
        }])

    except Exception as e:

        df = pd.DataFrame([{
            "archivo": file.name,
            "path": str(file),
            "tipo_geometria": None,
            "atributos": None,
            "estado": "ERROR",
            "error": str(e)
        }])

    df_rbind.append(df)

df_rbind = pd.concat(df_rbind, ignore_index=True)
```

```python

#carpeta = Path(r"E:/HYTSA/15 - HYTSA - C594 EL SALV DISEÑOS MAJAGÛE/00_Capas")
#carpeta = Path(r"E:/HYTSA/12 - HYTSA - C565 SAL  DISEÑO DE SANEAMIENTO Y AGUA POTABLE DESTINOS TURÍSTICOS/03_Datos")
carpeta = Path(r"E:/Luciano/00-Trabajos/00 - HYTSA/04 - HYTSA - El Salvador y Honduras")

df_rbind = []



for file in carpeta.rglob("*.gpkg"):

    print(file)

    try:

        # Obtener todas las capas dentro del GeoPackage
        capas = pyogrio.list_layers(file)

        for capa in capas:

            print("   └──", capa)

            try:
                # Leer la capa
                gdf = gpd.read_file(file, layer=capa)

                # Atributos
                atributos = gdf.columns.tolist()

                # Tipo de geometría
                tipo = gdf.geometry.geom_type.unique().tolist()

                df = pd.DataFrame([{
                    "archivo": file.name,
                    "capa": capa,
                    "path": str(file),
                    "tipo_geometria": tipo,
                    "atributos": atributos,
                    "estado": "OK",
                    "error": None
                }])

            except Exception as e:

                df = pd.DataFrame([{
                    "archivo": file.name,
                    "capa": capa,
                    "path": str(file),
                    "tipo_geometria": None,
                    "atributos": None,
                    "estado": "ERROR",
                    "error": str(e)
                }])

            df_rbind.append(df)

    except Exception as e:

        # Error al acceder al GeoPackage
        df = pd.DataFrame([{
            "archivo": file.name,
            "capa": None,
            "path": str(file),
            "tipo_geometria": None,
            "atributos": None,
            "estado": "ERROR",
            "error": str(e)
        }])

        df_rbind.append(df)


df_rbind = pd.concat(df_rbind, ignore_index=True)
```
